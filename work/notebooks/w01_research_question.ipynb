# Quick Look at the Data

The starter dataset contains **30,000 pages**. Around **54.2%** of the pages are labeled as declining. The correlation between **search_volume** and **impressions_90d** is **0.001**, showing that search volume alone is not a strong predictor of search visibility. These observations support studying content refresh prioritization.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print(f"Number of pages: {len(df)}")
print(f"Declining rate: {(df['trend_direction'].str.lower()=='down').mean():.3f}")
print(f"Correlation (search_volume vs impressions_90d): {df['search_volume'].corr(df['impressions_90d']):.3f}")
```
